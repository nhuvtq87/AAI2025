```python
import random


def estimate_sum_probability(target_sum, trials):
    """Estimate a two-dice sum probability using Monte Carlo simulation.

    Args:
        target_sum (int): Desired sum, from 2 through 12.
        trials (int): Positive number of simulated rolls.

    Returns:
        tuple: (simulated_probability, theoretical_probability, difference),
            where difference is simulated minus theoretical.

    Raises:
        ValueError: If either argument is not an integer (booleans excluded),
            target_sum is outside 2-12, or trials is not positive.
    """
    # Validate types explicitly; bool is otherwise considered an int in Python.
    if isinstance(target_sum, bool) or not isinstance(target_sum, int):
        raise ValueError("target_sum must be an integer from 2 to 12.")
    if not 2 <= target_sum <= 12:
        raise ValueError("target_sum must be an integer from 2 to 12.")

    if isinstance(trials, bool) or not isinstance(trials, int):
        raise ValueError("trials must be a positive integer.")
    if trials <= 0:
        raise ValueError("trials must be a positive integer.")

    matches = 0

    # Each trial rolls two independent, fair six-sided dice.
    for _ in range(trials):
        die_1 = random.randint(1, 6)
        die_2 = random.randint(1, 6)
        if die_1 + die_2 == target_sum:
            matches += 1

    simulated_probability = matches / trials

    # Favorable outcome counts rise from 1 to 6, then fall back to 1.
    favorable_outcomes = 6 - abs(7 - target_sum)
    theoretical_probability = favorable_outcomes / 36

    difference = simulated_probability - theoretical_probability
    return simulated_probability, theoretical_probability, difference


if __name__ == "__main__":
    # Example 1: The most likely sum.
    print('Example 1:',(estimate_sum_probability(7, 100000)))

    # Example 2: A boundary sum with only one favorable outcome.
    print('Example 2:',(estimate_sum_probability(2, 100000)))

    # Example 3: Catch and display an invalid-input error.
    try:
        print('Example 3:',(estimate_sum_probability(13, 100000)))
    except ValueError as error:
        print('Example 3:',(f"Error: {error}"))
```
